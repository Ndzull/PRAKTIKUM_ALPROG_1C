```python
# input list dengan loop input
input_list = []
while True:
    angka = input("Masukkan angka (atau ketik 'selesai' untuk berhenti): ")
    if angka.lower() == 'selesai':
        break
    try:
        input_list.append(int(angka))
    except ValueError:
        print("Harap masukkan angka yang valid.")
print("List angka:", input_list)
```
