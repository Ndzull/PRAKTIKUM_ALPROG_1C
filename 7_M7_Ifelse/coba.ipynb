```python
nilai = 90
if nilai >= 90:
    print("Predikat: Cumlaude")
elif(nilai >= 80):
    print("Predikat: Sangat Baik")
elif (nilai >= 70):
    print("Predikat: Baik")
else:
    print("Predikat: Cukup")
```

```python
match nilai:
    case nilai if nilai >= 90:
        print("Predikat: Cumlaude")
    case nilai if nilai >= 80:
        print("Predikat: Sangat Baik")
    case nilai if nilai >= 70:
        print("Predikat: Baik")
    case _:
        print("Predikat: Cukup")
```
